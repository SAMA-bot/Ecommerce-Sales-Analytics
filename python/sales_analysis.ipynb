# E-Commerce Sales Data Analysis

## Objective

Analyze e-commerce sales data to identify trends in sales,
customer segments, products, geographic performance,
and shipping modes.

## Tools

- Python
- Pandas
- NumPy
- Matplotlib
- Power BI

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("../data/train.csv")
```

```python
df.head()
```

```python
df.info()
```

```python
df["Order Date"] = pd.to_datetime(
    df["Order Date"],
    dayfirst=True
)

df["Ship Date"] = pd.to_datetime(
    df["Ship Date"],
    dayfirst=True
)
```

```python
df.info()
```

```python
df["Order Date"] = pd.to_datetime(df["Order Date"], dayfirst=True)
df["Ship Date"] = pd.to_datetime(df["Ship Date"], dayfirst=True)
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df = df.drop(columns=["Postal Code"])
```

```python
df.columns
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
print("Unique Orders:", df["Order ID"].nunique())
print("Unique Customers:", df["Customer ID"].nunique())
print("Unique Products:", df["Product ID"].nunique())
print("Categories:", df["Category"].nunique())
print("Sub-Categories:", df["Sub-Category"].nunique())
print("Regions:", df["Region"].nunique())
print("Segments:", df["Segment"].nunique())
print("Ship Modes:", df["Ship Mode"].nunique())
```

```python
df["Sales"].describe()
```

```python
print("Total Sales:", df["Sales"].sum())
```

```python
print("Average Sale per Row:", df["Sales"].mean())
```

```python
total_sales = df["Sales"].sum()
total_orders = df["Order ID"].nunique()

average_order_value = total_sales / total_orders

print(f"Total Sales: ${total_sales:,.2f}")
print(f"Total Orders: {total_orders:,}")
print(f"Average Order Value: ${average_order_value:,.2f}")
```

```python
df["Year"] = df["Order Date"].dt.year
df["Month"] = df["Order Date"].dt.month
df["Month Name"] = df["Order Date"].dt.month_name()
df["Year-Month"] = df["Order Date"].dt.to_period("M").astype(str)
df[["Order Date", "Year", "Month", "Month Name", "Year-Month"]].head()
```

```python
df["Sales"].describe()
```

## Sales By Category

```python
category_sales = df.groupby("Category")["Sales"].sum().sort_values(ascending = False)

category_sales
```

## Percentage Calculation

```python
category_percentage = (
    category_sales / category_sales.sum() * 100
)

category_percentage
```

```python
category_sales.plot(kind="bar")

plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)

plt.show()
```

## Sales By Sub Category

```python
subcategory_sales = df.groupby("Sub-Category")["Sales"].sum().sort_values(ascending=False)

subcategory_sales
```

```python
subcategory_sales.sort_values().plot(kind="barh")

plt.title("Sales by Sub-Category")
plt.xlabel("Total Sales")
plt.ylabel("Sub-Category")

plt.show()
```

```python
top_5_subcategories = subcategory_sales.head(5)

top_5_subcategories
```

## Customer Segment Analysis

```python
segment_sales = df.groupby("Segment")["Sales"].sum().sort_values(ascending=False)

segment_sales
```

```python
segment_sales.plot(kind="bar")

plt.title("Sales by Customer Segment")
plt.xlabel("Customer Segment")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)

plt.show()
```

## Regional Analysis

```python
region_sales = df.groupby("Region")["Sales"].sum().sort_values(ascending=False)

region_sales
```

```python
region_sales.plot(kind="bar")

plt.title("Sales by Region")
plt.xlabel("Region")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)

plt.show()
```

## Shipping Mode

```python
shipping_sales = df.groupby("Ship Mode")["Sales"].sum().sort_values(ascending=False)

shipping_sales
```

```python
shipping_sales.plot(kind="bar")

plt.title("Sales by Ship Mode")
plt.xlabel("Ship Mode")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)

plt.show()
```

## Sales over time

```python
monthly_sales = df.groupby("Year-Month")["Sales"].sum()

monthly_sales
```

```python
monthly_sales.plot(kind="line", figsize=(15, 5))

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)

plt.show()
```

## Top 10 Customers

```python
top_customers = df.groupby(["Customer ID", "Customer Name"])["Sales"].sum().sort_values(ascending=False).head(10)

top_customers
```

```python
top_customers.sort_values().plot(
    kind="barh",
    figsize=(11, 7)
)

plt.title("Top 10 Customers by Sales")
plt.xlabel("Total Sales")
plt.ylabel("Customer")

plt.show()
```

## Top 10 Products

```python
top_products = df.groupby("Product Name")["Sales"].sum().sort_values(ascending=False).head(10)

top_products
```

```python
top_products.sort_values().plot(
    kind="barh",
    figsize=(10, 6)
)

plt.title("Top 10 Products by Sales")
plt.xlabel("Total Sales")
plt.ylabel("Product")

plt.show()
```

## Bottom 10 Products

```python
bottom_products = df.groupby("Product Name")["Sales"].sum().sort_values(ascending=True).head(10)

bottom_products
```

```python
bottom_products.sort_values().plot(
    kind="barh",
    figsize=(10, 6)
)

plt.title("Bottom 10 Products by Sales")
plt.xlabel("Total Sales")
plt.ylabel("Product")

plt.show()
```

## State Analysis

```python
state_sales = df.groupby("State")["Sales"].sum().sort_values(ascending=False)

state_sales.head(10)
```

```python
state_sales.head(10).sort_values().plot(
    kind="barh",
    figsize=(10, 6)
)

plt.title("Top 10 States by Sales")
plt.xlabel("Total Sales")
plt.ylabel("State")

plt.show()
```

```python
state_sales.tail(10)
```

## City Analysis

```python
city_sales = df.groupby("City")["Sales"].sum().sort_values(ascending=False)

city_sales.head(10)
```

```python
city_sales.head(10).sort_values().plot(
    kind="barh",
    figsize=(10, 6)
)

plt.title("Top 10 States by Sales")
plt.xlabel("Total Sales")
plt.ylabel("State")

plt.show()
```

```python
best_month = monthly_sales.idxmax()
best_month_sales = monthly_sales.max()

print("Best Month:", best_month)
print(f"Sales: ${best_month_sales:,.2f}")
```

```python
worst_month = monthly_sales.idxmin()
worst_month_sales = monthly_sales.min()

print("Worst Month:", worst_month)
print(f"Sales: ${worst_month_sales:,.2f}")
```

## Annual Sales

```python
yearly_sales = df.groupby("Year")["Sales"].sum().sort_index()

yearly_sales
```

```python
yearly_sales.plot(
    kind="bar",
    figsize=(8, 5)
)

plt.title("Yearly Sales")
plt.xlabel("Year")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)

plt.show()
```

## Calculate year-over-year growth

```python
yearly_growth = yearly_sales.pct_change() * 100

yearly_growth
```

```python
yearly_growth = yearly_growth.round(2)

yearly_growth
```

## Average Sales Per Customer

```python
total_customers = df["Customer ID"].nunique()

print(f"Total Customers: {total_customers:,}")
```

```python
average_sales_per_customer = total_sales / total_customers

print(f"Average Sales per Customer: ${average_sales_per_customer:,.2f}")
```

```python
average_orders_per_customer = total_orders / total_customers

print(f"Average Orders per Customer: {average_orders_per_customer:.2f}")
```

```python
top_10_customer_sales = top_customers.sum()

top_10_customer_percentage = (
    top_10_customer_sales / total_sales
) * 100

print(f"Top 10 Customer Sales: ${top_10_customer_sales:,.2f}")
print(f"Top 10 Customer Contribution: {top_10_customer_percentage:.2f}%")
```

```python
top_10_product_sales = top_products.sum()

top_10_product_percentage = (
    top_10_product_sales / total_sales
) * 100

print(f"Top 10 Product Sales: ${top_10_product_sales:,.2f}")
print(f"Top 10 Product Contribution: {top_10_product_percentage:.2f}%")
```
